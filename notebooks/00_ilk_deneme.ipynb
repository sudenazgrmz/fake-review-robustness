```python
import pandas as pd
import matplotlib.pyplot as plt

# Küçük bir örnek tablo (gerçek veri değil, sadece alıştırma)
veri = pd.DataFrame({
    "yorum": ["Harika ürün", "Berbat", "Fena değil", "Çok iyi, tavsiye ederim", "Hiç beğenmedim"],
    "etiket": ["gerçek", "gerçek", "gerçek", "sahte", "sahte"],
})

print(veri)
print(veri["etiket"].value_counts())

veri["etiket"].value_counts().plot(kind="bar")
plt.title("Etiket dağılımı")
plt.show()
```
