```python
import pandas as pd

df = pd.read_csv("deceptive-opinion.csv")
print(df.shape)
print(df.columns.tolist())
print(df.head(3))
```

```python
# 1) Sınıf dağılımı: etiket ve polarite
print(pd.crosstab(df["deceptive"], df["polarity"]))
print()

# 2) Kaynak: sahte ve gerçek yorumlar nereden gelmiş?
print(pd.crosstab(df["deceptive"], df["source"]))
print()

# 3) Ortalama uzunluk (kelime sayısı)
df["kelime"] = df["text"].str.split().str.len()
print(df.groupby("deceptive")["kelime"].agg(["mean", "median", "min", "max"]).round(1))
print()

# 4) Kaç otel var, otel başına kaç yorum?
print(df["hotel"].nunique(), "otel")
print(df["hotel"].value_counts().head())
```

```python
ornek = df.sample(6, random_state=7)[["text", "deceptive"]].reset_index(drop=True)
for i, m in enumerate(ornek["text"]):
    print(f"--- Yorum {i+1} ---")
    print(m[:500], "\n")
```
